# LangChain Beginner Guide
A step-by-step introduction to using LangChain with OpenAI models.

## 1. Install LangChain
We install the library first. `-Uq` means upgrade quietly.

```python
!pip install -Uq langchain[openai]
```

## 2. Import Tools and Set API Key
We import Python modules and securely load your OpenAI key from Colab Secrets.

```python

from langchain.chat_models import init_chat_model
import os
from google.colab import userdata

# Reads a secret named 'openai' from Colab Secrets
os.environ["OPENAI_API_KEY"] = userdata.get("openai")
```

### Line-by-line explanation
- `init_chat_model`: helper that creates a chat model.
- `os.environ[...]`: stores the API key so LangChain can use it.
- `userdata.get(...)`: reads a hidden secret safely.

## 3. Create a Model

```python
model = init_chat_model(model="gpt-4.1-nano")
```

This creates a lightweight chat model object. No question is asked yet.

## 4. Ask the Model a Question

```python
response = model.invoke("What is the capital of Maharashtra?")
response
```

`invoke()` sends one request and returns an AI message object.

```python
print(response.content)
```

`.content` gives only the text answer.

## 5. View Metadata

```python
from pprint import pprint
pprint(response.response_metadata)
```

Metadata may include model name, token usage, finish reason, etc.

## 6. Change Model Creativity

```python

model = init_chat_model(
    model="gpt-4.1-nano",
    temperature=1.0,
    max_tokens=50
)

response = model.invoke("Write a fun slogan on LLM.")
print(response.content)
```

Higher `temperature` usually means more creative and varied responses.

## 7. Create an Agent
An agent can reason and work with messages/tools.

```python

from langchain.agents import create_agent

agent = create_agent(model=model)
```

## 8. Send a Human Message

```python

from langchain.messages import HumanMessage

response = agent.invoke({
    "messages": [
        HumanMessage(content="What is the capital of Japan?")
    ]
})
```

```python
from pprint import pprint
pprint(response)
```

The result often contains a `messages` list with the conversation history.

```python
pprint(response['messages'][-1].content)
```

`[-1]` means the last message, usually the latest AI reply.

## 9. Multi-turn Conversation Memory

```python

from langchain.messages import AIMessage

response = agent.invoke({
    "messages": [
        HumanMessage(content="What's the capital of the Moon?"),
        AIMessage(content="The capital of the Moon is Luna City."),
        HumanMessage(content="Interesting, tell me more about Luna City.")
    ]
})

pprint(response)
```

### Why this matters
The agent uses previous messages as context. If earlier messages contain fictional information, the model may continue that story.

## Beginner Tips
1. Start with `model.invoke()` before agents.
2. Use `.content` for plain text answers.
3. Keep API keys secret.
4. Use agents when you need tools or multi-step workflows.

#Basic Prompting

Below are some examples on how to use prompt to get better output

```python
from langchain.agents import create_agent
from langchain.messages import HumanMessage

question = HumanMessage(content="Tell me about Mumbai")

system_prompt = "You are a Travel Blogger. Write few interesting facts about the City"

scifi_agent = create_agent(
    model="gpt-4.1-nano",
    system_prompt=system_prompt
)

response = scifi_agent.invoke(
    {"messages": [question]}
)

print(response['messages'][1].content)
```

## Structured prompts

```python
from langchain.messages import HumanMessage

question = HumanMessage(content="Tell me about Mumbai")

system_prompt = """

You are a Travel Blogger. Write few interesting facts about the City.

Please keep to the below structure.

Name: The name of the capital city

Location: Where it is based

Vibe: 2-3 words to describe its vibe

Economy: Main industries

Tourists Spots: 2-3 tourist spots

Politcal Facts: 2-3 facts about its politics

Transportatons: 2-3 modes of transportation

GDP: Gross Domestic Product

Number Of billionaires and millionaires: Number of billionaires based on the fact dates

"""

scifi_agent = create_agent(
    model="gpt-4.1-nano",
    system_prompt=system_prompt
)

response = scifi_agent.invoke(
    {"messages": [question]}
)

print(response['messages'][1].content)
```

## Structured Output

We install the library call pydantic to create a class (OOP)

```python
pip install pydantic
```

```python
from pydantic import BaseModel

class CityInfo(BaseModel):
    name: str
    location: str
    vibe: str
    economy: str
    tourist_spots: str
    politics: str
    transportations: str
    gdp: str
    billionaires: str

scifi_agent = create_agent(
model="gpt-4.1-nano",
system_prompt=system_prompt,
response_format=CityInfo
)

response = scifi_agent.invoke(
    {"messages": [question]}
)

print(response['structured_response'])
```

# Using Tools

## Defining 2 tools method which will be called by agent

```python
from langchain.tools import tool

@tool("add_number",description="Calculate Additional Of Two Numbers")
def addition(a: float , b: float)-> int:
  """ Calculate the sum of two numbers """
  return int(a) + int(b)
```

```python
@tool("substract_number",description="Calculate the difference of two numbers")
def substraction(a: float , b: float):
  """ Calculate the difference of two numbers """
  return int(a) - int(b)
```

## Invoke tool manually to see the output\

```python
addition.invoke({"a":10,"b":20})

```

```python
substraction.invoke({"a":40,"b":20})
```

## Create an agent and provided tools which can be run by the agent as an Array

```python
from langchain.agents import create_agent

agent = create_agent(model="gpt-4.1-nano",tools=[addition,substraction],
                     system_prompt="You are a math teacher. Use your tools to calculate addition or substraction of any given number.Any Question outside this context, please response politely saying I can only provide you help with addition and substraction")
```

## Pass user input and see the response

```python
from langchain.messages import HumanMessage

question = HumanMessage("100 + 200")
response = agent.invoke({"messages":[question]})
print(response['messages'][-1].content)
```

```python
from pprint import pprint
pprint(response['messages'])
```
